```python
import mlflow
import sys
sys.path.append('..')

from src.model import load_blip_model

mlflow.set_tracking_uri("sqlite:///../mlflow.db")
mlflow.set_experiment("flickr8k-image-captioning")

processor, model, device = load_blip_model()
print("Model loaded successfully!")
```

```python
with mlflow.start_run(run_name="registered_best_baseline") as run:
    mlflow.log_param("model", "Salesforce/blip-image-captioning-base")
    mlflow.log_param("decoding_strategy", "beam_search")
    mlflow.log_param("num_beams", 3)
    mlflow.log_param("learning_rate", 5e-5)
    mlflow.log_metric("val_bleu", 0.284)

    model.save_pretrained("../data/processed/registered_model")
    processor.save_pretrained("../data/processed/registered_model")
    mlflow.log_artifacts("../data/processed/registered_model", artifact_path="blip_captioning_model")

    run_id = run.info.run_id
    print(f"Model artifacts logged! Run ID: {run_id}")

# Register it separately
result = mlflow.register_model(
    model_uri=f"runs:/{run_id}/blip_captioning_model",
    name="flickr8k-blip-captioning-baseline"
)
print(f"Registered as version {result.version}")
```
